```python
import os
os.environ['TF_CPP_MIN_LOG_LEVEL']='2'
import tensorflow as tf
```

```python
X = tf.placeholder(tf.float32, [None, 3])
print(X)
```

```python
x_data = [[1, 2, 3], [4, 5, 6]]
```

```python
W = tf.Variable(tf.random_normal([3, 2]))
b = tf.Variable(tf.random_normal([2]))
```

```python
expr = tf.matmul(X, W) + b
expr_simple = tf.matmul(X, W)
```

```python
sess = tf.Session()
sess.run(tf.global_variables_initializer())
```

```python
print("=== x_data ===")
print(x_data)
print("=== W ===")
print(sess.run(W))
print("=== b ===")
print(sess.run(b))
print("=== expr ===")
print(sess.run(expr, feed_dict = {X: x_data}))
print("=== expr_simple ===")
print(sess.run(expr_simple, feed_dict = {X: x_data}))
```

```python
sess.close()
```

```python
import os
os.environ['TF_CPP_MIN_LOG_LEVEL']='2'
import tensorflow as tf
```

```python
hello = tf.constant("Hello, TensorFlow!")
print(hello)
```

```python
a = tf.constant(10)
b = tf.constant(32)
c = tf.add(a, b)
print(c)
```

```python
sess = tf.Session()
```

```python
print(sess.run(hello))
print(sess.run([a, b, c]))
```

```python
sess.close()
```

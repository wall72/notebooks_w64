```python
import datetime
from pandas_datareader import data, wb
import matplotlib.pyplot as plt
```

```python
def download_stock_data(file_name, company_code, year1, month1, date1, year2, month2, date2):
    start = datetime.datetime(year1, month1, date1)
    end = datetime.datetime(year2, month2, date2)
    
    df = data.DataReader(
        "KRX:%s" % (company_code),
        "google",
        start,
        end
    )
    
    df.to_csv(file_name, encoding='utf-8')
    
    return df
```

```python
company_code = "005930"
file_name = "samsung.csv"
```

```python
df = download_stock_data(file_name, company_code, 2017, 1, 1, 2017, 6, 30)
```

```python
plt.plot(df["Close"])
plt.grid()
plt.title("%s Chart" % (company_code))
plt.xlabel("date")
plt.ylabel("price")
plt.show()
```

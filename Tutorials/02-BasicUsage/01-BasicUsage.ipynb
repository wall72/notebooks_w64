## Building the graph

```python
import tensorflow as tf

# 1x2 행렬을 만드는 constant op 생성
# 생성된 op는 default graph 에 노드로 추가됨

# 컨스트럭터에서 나온 값은 constant op의 결과 값
matrix1 = tf.constant([[3., 3.]])

# 2x1 행렬을 추가로 만듦
matrix2 = tf.constant([[2.], [2.]])

# 'matrix1, 'matrix2를 입력으로 matmul op 생성
# 결과적으로 행렬곱의 결과가 'product'에 저장됨
product = tf.matmul(matrix1, matrix2)

# default graph 는 세개의 노드로 구성됨
# 두개의 constant() op와 하나의 matmul() op
```

## Launching the default graph

```python
# default graph 실행
sess = tf.Session()

# matmul op 를 실행하기 위해 session의 'run()'메소드를 호출
# matmul op 의 결과값을 표현하는 'product'를 인자로 넘김
# 이렇게 메소드를 실행하면 matmul op의 결과를 얻게됨
# op의 모든 입력은 session 에서 병렬로 자동실행
# 'run(product)'가 실해되면 세개의 op가 그래프에서 실행됨
result = sess.run(product)
print(result)
# ==> [[12.]]

# 실행을 종료하기 위해 session을 닫음
sess.close()
```

### 'with' 블록 사용

```python
# 'with' 블록을 사용하면 자동으로 'close()'가 실행됨
with tf.Session() as sess:
    result = sess.run(product)
    print(result)
```

### 'devide()' 메소드를 사용해 op 실행 하드웨어를 특정
+ "/cpu:0": 컴퓨터의 CPU.
+ "/gpu:0": 컴퓨터의 1번째 GPU.
+ "/gpu:1": 컴퓨터의 2번쨰 GPU.

```python
with tf.Session() as sess:
    with tf.device("/cpu:0"):
        matrix1 = tf.constant([[3., 3.]])
        matrix2 = tf.constant([[2.],[2.]])
        product = tf.matmul(matrix1, matrix2)
        result = sess.run(product)
        print(result)
```

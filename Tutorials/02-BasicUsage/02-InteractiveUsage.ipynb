```python
import tensorflow as tf

# Interactive session 을 시작
sess = tf.InteractiveSession()

x = tf.Variable([1.0, 2.0])
a = tf.constant([3.0, 3.0])

# 'x' 변수를 초기화 하기 위해 initializer op 의 'run()' 메소드를 호출
x.initializer.run()

#  x - a 를 수행하는 op를 추가하고 실행
sub = tf.sub(x, a)
print(sub.eval())
# ==> [-2. -1.]

# 실행을 종료하기 위해 session을 닫음
sess.close()
```

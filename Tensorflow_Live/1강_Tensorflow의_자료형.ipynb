## 1강 : Tensorflow 자료형

```python
import tensorflow as tf
```

```python
# placeholder
ph = tf.placeholder(dtype=tf.float32, shape=[3,3])

# variable
var = tf.Variable([1, 2, 3, 4, 5], dtype=tf.float32)

# constant
const = tf.constant([10, 20, 30, 40, 50], dtype=tf.float32)

print(ph)
print(var)
print(const)
```

```python
a = tf.placeholder(dtype=tf.float32)
b = tf.placeholder(dtype=tf.float32)
c = tf.placeholder(dtype=tf.float32)

d = a * b + c

sess = tf.Session()
result = sess.run(d, feed_dict = {a: 5, b: 2, c: 3})
print(result)
```
